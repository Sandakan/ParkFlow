```python
# Load the base model (v8s is recommended for a balance of speed/accuracy)
model = YOLO("yolo26n.pt")
PROJECT_DIR = LOCAL_DATASET


# Start Training
# Results are saved directly to your Drive
# Corrected path: data.yaml is inside 'yolo_dataset' subdirectory
corrected_data_path = os.path.join(DATASET_PATH, "data.yaml")
print(f"Attempting to train with data.yaml at: {corrected_data_path}")
model.train(
    data=corrected_data_path,
    epochs=50,
    imgsz=640,
    batch=16,
    project=f"{PROJECT_DIR}/training_runs",
    name="pklot_base_model",
    amp=False,
    device=device
)
```
